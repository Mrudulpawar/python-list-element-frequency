## task1.
##### ->WAP to get unique elements list from given list without changing data position...

```python
l=[5,4,5,4,6,7,8,4,3,5,3,4]
unique_list=[]
for data in l:
    if data not in unique_list:
        unique_list.append(data)
print(unique_list)
```

## task2.
##### -> find freq of every elements from above input list in dictionary...


```python
l = [5,4,5,4,6,7,8,4,3,5,3,4]
d = {}
for i in l:
    d[i] = l.count(i)
print(d)
```
